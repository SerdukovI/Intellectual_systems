## 9. Перевод даты в формат date

```python

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Order_Time'] = pd.to_datetime(df['Order_Time'], format='%H:%M:%S', errors='coerce').dt.time
```

## 10. Изменение регистра,удаление пробелов, замена на нормальный none

```python
text_columns = [
    'Customer_Name', 'Customer_Gender', 'Customer_Segment',
    'Sales_Channel', 'Country', 'Region', 'City',
    'Product_Name', 'Product_Category', 'Product_Subcategory',
    'Payment_Method', 'Order_Status', 'Shipping_Method',
    'Return_Flag', 'Return_Reason', 'Sales_Representative',
    'Promotion_Code'
]
for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.lower().str.strip()
for col in text_columns:
    if col in df.columns:
        df[col] = df[col].replace(['nan', 'none', 'null', ''], pd.NA)
print("Уникальные значения в Customer_Gender после очистки:")
print(df['Customer_Gender'].unique())
print("\nУникальные значения в City после очистки (первые 10):")
print(df['City'].unique()[:10])
print("\nУникальные значения в Product_Category после очистки:")
print(df['Product_Category'].unique())
print("\nИнформация о типах данных:")
print(df[['Order_Date', 'Order_Time', 'Customer_Gender', 'City']].dtypes)
```

## 11. Дубликаты !

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

## 12. Чистим аномальные значения от отрицательных чисел в количестве

```python
print(f"Строк с отрицательным Quantity до: {len(df[df['Quantity'] < 0])}")
df.loc[df['Quantity'] < 0, 'Quantity'] = 1
print(f"Строк с отрицательным Quantity после: {len(df[df['Quantity'] < 0])}")
```

## 13. Чистим аномальные значения возраста

```python
median_age = df.loc[df['Customer_Age'] <= 90, 'Customer_Age'].median()
print(f"Максимальный возраст до: {df['Customer_Age'].max()}")
df.loc[df['Customer_Age'] > 90, 'Customer_Age'] = median_age
df.loc[df['Customer_Age'] < 18, 'Customer_Age'] = median_age
print(f"Максимальный возраст после: {df['Customer_Age'].max()}")
```

## 14. Округление значений

```python
df['Unit_Price'] = df['Unit_Price'].round(2)
```

## 15. Аномалия со скидкой

```python
invalid_discounts = df[(df['Discount_Percentage'] < 0) | (df['Discount_Percentage'] > 100)]
print(f"Строк с некорректной скидкой: {len(invalid_discounts)}")
df = df[(df['Discount_Percentage'] >= 0) & (df['Discount_Percentage'] <= 100)]
print(f"Новый размер датасета: {df.shape}")
print(f"Максимальная скидка после очистки: {df['Discount_Percentage'].max()}")
```

## 16. Проверка отрицательных сумм продажи (возвраты)

```python
negative_sales = df[df['Sales_Amount'] < 0]
print(f"Строк с отрицательной суммой продажи (возвраты): {len(negative_sales)}")
```

## 17. Промежуточная статистика

```python
print(f"Диапазон дат: от {df['Order_Date'].min()} до {df['Order_Date'].max()}")
print(f"Общее количество строк: {len(df)}")
print(df[['Quantity', 'Customer_Age', 'Unit_Price', 'Discount_Percentage']].describe().round(2))
```

## 18. Заполняем пропуски с помощью средних значений в столбцах

```python
numeric_cols_to_fill = ['Customer_Age', 'Customer_Rating', 'Inventory_Level', 'Delivery_Days']
for col in numeric_cols_to_fill:
    if col in df.columns:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"Столбец '{col}': пропуски заполнены медианой ({median_val:.2f})")
```

## 19. Удаляем аномалии с возвратом товаров

```python
rows_to_drop = len(df[(df['Return_Flag'] == 'yes') & (df['Return_Reason'].isna())])
df = df[~((df['Return_Flag'] == 'yes') & (df['Return_Reason'].isna()))]
print(f"Удалено строк (возврат есть, причина не указана): {rows_to_drop}")
```

## 20. Заменяем отсутствующие значения в столбце Промо

```python
df['Promotion_Code'] = df['Promotion_Code'].fillna('No Promo')
print("Столбец 'Promotion_Code': пропуски заполнены 'No Promo'")
```

## 21. Заполняем пропуски модой в столбце пола

```python
if 'Customer_Gender' in df.columns:
    mode_gender = df['Customer_Gender'].mode()[0]
    df['Customer_Gender'] = df['Customer_Gender'].fillna(mode_gender)
    print(f"Столбец 'Customer_Gender': пропуски заполнены модой ({mode_gender})")
```

## 22. Делаем все тоже самое для остальных проблемных столбцов

```python
other_cat_cols = ['Payment_Method', 'Order_Status', 'Shipping_Method', 'Sales_Representative']
for col in other_cat_cols:
    if col in df.columns:
        mode_val = df[col].mode()[0]
        df[col] = df[col].fillna(mode_val)
        print(f"Столбец '{col}': пропуски заполнены модой ({mode_val})")
```

## 23. Промежуточный результат про пропускам

```python
print("Оставшиеся пропуски по столбцам:")
missing_after = df.isnull().sum()
print(missing_after[missing_after > 0])
if missing_after.sum() == 0:
    print("\nВсе пропуски успешно заполнены.")
else:
    print(f"\nОсталось пропусков: {missing_after.sum()}.")
```

## 24. Удаляем строки с пропущенными значениями в данных столбцах

```python
rows_before = len(df)
df = df.dropna(subset=['Customer_Name', 'Product_Name'])
rows_after = len(df)
print(f"Удалено строк с пропущенными именами: {rows_before - rows_after}")
```

## 25. Поиск дубликатов

```python
full_duplicates = df.duplicated().sum()
print(f"\nПолных дубликатов: {full_duplicates}")
id_duplicates = df.duplicated(subset=['Transaction_ID']).sum()
print(f"Дубликатов по Transaction_ID: {id_duplicates}")
```

## 26. Удаляем дубликаты 

```python
if full_duplicates > 0:
    df = df.drop_duplicates()
    print(f"Осталось строк: {len(df)}")
if id_duplicates > 0:
    df = df.drop_duplicates(subset=['Transaction_ID'], keep='first')
    print(f"Осталось строк: {len(df)}")
```

## 27. Удаление столбцов 

```python
columns_to_drop_final = [
    'Cost_Amount', 
    'Profit',
    'Customer_ID', 
    'Product_ID', 
    'Store_ID', 
    'Sales_Representative',
    'Order_Year', 
    'Return_Reason', 
    'Order_Time'
    'Transaction_ID',
    'Customer_Name',
    'Product_Name',
    'Store_Name',
    'Country',
    'Region',           
    'Order_ID',       
    'Promotion_Code',
    'Order_Status',
    'Inventory_Level',
    'Order_Time',
    'Delivery_Days'
]
columns_to_drop_final = [col for col in columns_to_drop_final if col in df.columns]
df = df.drop(columns=columns_to_drop_final)
print(f"Удалены столбцы: {columns_to_drop_final}")
print(f"Новый размер таблицы: {df.shape}")
print("\nОставшиеся столбцы:")
print(df.columns.tolist())
```

## 2.2 Начинаем удалять выбросы с проблемных столбцов

## Для Quantity

```python
# Определяем нижнюю и верхнюю границы для выбросов, используя межквартильный размах (IQR)
Q1 = df['Quantity'].quantile(0.25)  # Первый квартиль (25% данных)
Q3 = df['Quantity'].quantile(0.75)  # Третий квартиль (75% данных)
IQR = Q3 - Q1  # Межквартильный размах (разница между Q3 и Q1)
# Вычисляем нижнюю и верхнюю границы для выбросов
lower_bound = Q1 - 0.8 * IQR  # Нижняя граница
upper_bound = Q3 + 0.8 * IQR  # Верхняя граница
# Фильтруем данные
df = df[(df['Quantity'] >= lower_bound) & (df['Quantity'] <= upper_bound)]
# Строим boxplot
quantityBoxplot = sns.boxplot(x=df['Quantity'])
quantityBoxplot.figure.set_size_inches(10, 8)
plt.show()
```

## Для Unit_Price

```python
Q1 = df['Unit_Price'].quantile(0.25)
Q3 = df['Unit_Price'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 0.8 * IQR
upper_bound = Q3 + 0.8 * IQR
df = df[(df['Unit_Price'] >= lower_bound) & (df['Unit_Price'] <= upper_bound)]
quantityBoxplot = sns.boxplot(x=df['Unit_Price'])
quantityBoxplot.figure.set_size_inches(10, 8)
plt.show()
```

## Для Discount_Percentage

```python
Q1 = df['Discount_Percentage'].quantile(0.25)
Q3 = df['Discount_Percentage'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 0.8 * IQR
upper_bound = Q3 + 0.8 * IQR
df = df[(df['Discount_Percentage'] >= lower_bound) & (df['Discount_Percentage'] <= upper_bound)]
quantityBoxplot = sns.boxplot(x=df['Discount_Percentage'])
quantityBoxplot.figure.set_size_inches(10, 8)
plt.show()
```
